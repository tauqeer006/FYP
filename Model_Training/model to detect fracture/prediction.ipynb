```python
from ultralytics import YOLO

model = YOLO("best.pt")  
```

```python
results = model("E:/Tauqeer Projects/Tauqeer And Ahtisham FYP/Dataset/Images/0003_0662359351_01_WRI-R2_M011.png")  

results[0].show()
```

```python
import shap
import cv2
import torch

image_path = "E:/Tauqeer Projects/Tauqeer And Ahtisham FYP/Dataset/Images/0003_0662359351_01_WRI-R2_M011.png"
image = cv2.imread(image_path)
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
image_tensor = torch.tensor(image).permute(2, 0, 1).unsqueeze(0).float() / 255.0

explainer = shap.Explainer(model.model)
shap_values = explainer(image_tensor)

shap.image_plot(shap_values)
```
