```python
import numpy as np
import mediapipe as mp
import cv2
import os
import pickle
```

```python
dataset = "dataset"
OUTPUT_FILE = "pose_dataset.pkl"
```

```python
mp_pose = mp.solutions.pose
```

```python
mp_pose
```

```python
pose = mp_pose.Pose(static_image_mode=False,
                    model_complexity=1,
                    enable_segmentation=False,
                    min_detection_confidence=0.5,
                    min_tracking_confidence=0.5)
```

```python
def extract_keypoints_from_video(video_path):
    """
    Extract pose landmarks from a video.
    Returns a numpy array of shape (frames, 132)
    """
    cap = cv2.VideoCapture(video_path)
    sequence = []

    while cap.isOpened():
        ret, frame = cap.read()
        if not ret:
            break

        frame_rgb = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
        results = pose.process(frame_rgb)

        if results.pose_landmarks:
            landmarks = []
            for lm in results.pose_landmarks.landmark:
                landmarks.extend([lm.x, lm.y, lm.z, lm.visibility])
            sequence.append(landmarks)

    cap.release()
    return np.array(sequence, dtype=np.float32)

```

```python
all_sequences = []
video_files = [f for f in os.listdir(dataset) if f.endswith((".mp4", ".avi", ".mov"))]


```

```python
len(video_files)
```

```python
for idx, file in enumerate(video_files):
    video_path = os.path.join(dataset, file)
    print(f"[{idx+1}/{len(video_files)}] Processing {file} ...")

    seq = extract_keypoints_from_video(video_path)
    if seq.size > 0:
        all_sequences.append(seq)

with open(OUTPUT_FILE, "wb") as f:
    pickle.dump(all_sequences, f)

print(f"✅ Saved {len(all_sequences)} sequences to {OUTPUT_FILE}")
```
